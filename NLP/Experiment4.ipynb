```python
import spacy

nlp = spacy.load("en_core_web_sm")

print("Model loaded successfully!")
```

```python
import nltk

nltk.download("reuters")
```

```python
from nltk.corpus import reuters

print(len(reuters.fileids()))
file_id = reuters.fileids()[0]

text = reuters.raw(file_id)

print(text)
```

```python
doc = nlp(text)

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

```python
from spacy import displacy

displacy.render(
    doc,
    style="ent",
    jupyter=True
)
```

```python
print("PERSON entities:")

for ent in doc.ents:
    if ent.label_ == "PERSON":
        print(ent.text)
```

```python
import pandas as pd

entities = []

for ent in doc.ents:
    entities.append({
        "Entity": ent.text,
        "Type": ent.label_,
        "Description": spacy.explain(ent.label_)
    })

entity_df = pd.DataFrame(entities)

display(entity_df)
```

```python
medical_text = """
The patient was diagnosed with diabetes and hypertension.
The doctor prescribed Metformin and Aspirin.
The patient returned to Pune for a follow-up examination.
"""
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

ruler = nlp.add_pipe("entity_ruler", before="ner")

patterns = [
    {"label": "DISEASE", "pattern": "diabetes"},
    {"label": "DISEASE", "pattern": "hypertension"},
    {"label": "MEDICATION", "pattern": "Metformin"},
    {"label": "MEDICATION", "pattern": "Aspirin"},
    {"label": "GPE", "pattern": "Pune"}
]

ruler.add_patterns(patterns)

medical_doc = nlp(medical_text)

for ent in medical_doc.ents:
    print(ent.text, "->", ent.label_)
```
