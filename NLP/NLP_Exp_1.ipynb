```python
# ============================================================
# TASK 1: Text Preprocessing & Frequency Analysis of News Articles
# Run this cell as-is in Google Colab.
# ============================================================
!pip install nltk -q
import nltk, re, string
from collections import Counter
# Download required NLTK resources (safe to re-run)
for resource in ['punkt', 'punkt_tab', 'stopwords', 'wordnet', 'omw-1.4']:
    try:
        nltk.download(resource, quiet=True)
    except Exception as e:
        print(f"Note: could not download {resource} ({e})")
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
# --------------------------------------------------------
# Sample dataset of news articles (swap in your own list/CSV)
# --------------------------------------------------------
news_articles = [
    "The stock market rallied today as investors reacted positively to the "
    "Federal Reserve's decision. Analysts say the rally could continue "
    "into next week!",
    "Scientists have discovered a new exoplanet that may support life. "
    "The discovery, published today in a leading journal, has excited "
    "the global research community.",
    "The government announced new policies aimed at reducing carbon "
    "emissions by 2030. Environmentalists welcomed the move, but "
    "industry leaders raised concerns about costs.",
    "In a thrilling final match, the national football team won the "
    "championship after a dramatic penalty shootout. Fans celebrated "
    "wildly across the country.",
    "Tech companies are racing to develop advanced artificial intelligence "
    "systems. Experts warn that regulation must keep pace with rapid "
    "innovation in the industry."
]
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

```python
def clean_and_process(text):
    """Lowercase, tokenize, strip punctuation/stopwords, stem & lemmatize."""
    # d. Normalize -> lowercase
    text_lower = text.lower()
    # b. Sentence tokenization
    sentences = sent_tokenize(text_lower)
    # a. Remove punctuation & special characters
    text_no_punct = text_lower.translate(str.maketrans('', '', string.punctuation))
    text_clean = re.sub(r'[^a-z0-9\s]', ' ', text_no_punct)
    # b. Word tokenization
    word_tokens = word_tokenize(text_clean)
    word_tokens = [w for w in word_tokens if w.isalpha() and len(w) > 1]
    # a. Remove stop words
    tokens_no_stopwords = [w for w in word_tokens if w not in stop_words]
    # c. Stemming
    stemmed_tokens = [stemmer.stem(w) for w in tokens_no_stopwords]
    # c. Lemmatization
    lemmatized_tokens = [lemmatizer.lemmatize(w) for w in tokens_no_stopwords]
    return {
        'sentences': sentences,
        'tokens': tokens_no_stopwords,
        'stemmed': stemmed_tokens,
        'lemmatized': lemmatized_tokens
    }
all_lemmatized_tokens = []
```

```python
for idx, article in enumerate(news_articles, start=1):
    result = clean_and_process(article)
    print(f"\n--- Article {idx} ---")
    print("Sentences        :", result['sentences'])
    print("Tokens (no stop) :", result['tokens'])
    print("Stemmed tokens   :", result['stemmed'])
    print("Lemmatized tokens:", result['lemmatized'])
    all_lemmatized_tokens.extend(result['lemmatized'])
# --------------------------------------------------------
# Frequency analysis across all articles (using lemmas)
# --------------------------------------------------------
freq_dist = Counter(all_lemmatized_tokens)
print("\n=== Top 15 Most Frequent Terms Across All Articles ===")
for word, count in freq_dist.most_common(15):
    print(f"{word:15s} : {count}")
```
